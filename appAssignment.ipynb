## 03 - Regex Matching

```python
s = eval(input().split("=", 1)[-1])
p = eval(input().split("=", 1)[-1])
m, n = len(s), len(p)
d = [[False] * (n + 1) for _ in range(m + 1)]
d[0][0] = True
for j in range(2, n + 1):
    if p[j - 1] == "*":
        d[0][j] = d[0][j - 2]
for i in range(1, m + 1):
    for j in range(1, n + 1):
        if p[j - 1] == "." or p[j - 1] == s[i - 1]:
            d[i][j] = d[i - 1][j - 1]
        elif p[j - 1] == "*":
            d[i][j] = d[i][j - 2]
            if p[j - 2] == "." or p[j - 2] == s[i - 1]:
                d[i][j] |= d[i - 1][j]
print(str(d[m][n]).lower())
```

## 04 - Decode Ways

```python
s = input().split("=", 1)[-1].strip().strip('"')
r = []
def gen(i, x):
    if i == len(s):
        r.append(x)
        return
    if s[i] != "0":
        gen(i + 1, x + chr(int(s[i]) + 64))
    if i + 1 < len(s) and s[i] != "0" and 10 <= int(s[i:i+2]) <= 26:
        gen(i + 2, x + chr(int(s[i:i+2]) + 64))
gen(0, "")
print("Total Ways:", len(r))
print("Decoded Strings:")
if r:
    for x in r:
        print(x)
else:
    print("None")
```

## 05 - Balanced Binary Tree

```python
import ast
a = ast.literal_eval(input().split("=", 1)[-1])
def height(i):
    if i >= len(a) or a[i] is None:
        return 0
    l = height(2 * i + 1)
    r = height(2 * i + 2)
    if l == -1 or r == -1 or abs(l - r) > 1:
        return -1
    return max(l, r) + 1
print(str(height(0) != -1).lower())
```

## 07 - Minimize Consecutive Zeros

```python
from itertools import combinations
L, K = map(int, input().split())
r = []
for p in combinations(range(L), K):
    s = ["0"] * L
    for i in p:
        s[i] = "1"
    s = "".join(s)
    m = max(map(len, s.split("1")))
    r.append((s, m))
print("Possible Strings:")
for s, m in r:
    print(s, "→ max zeros =", m)
v = min(m for s, m in r)
a = [s for s, m in r if m == v]
print("Optimal Arrangement(s):", ", ".join(a))
print("Minimum Longest Zero Block:", v)
```

## 19 - Smallest Sufficient Team

```python
import ast
skills = ast.literal_eval(input())
people = ast.literal_eval(input())
n = len(skills)
idx = {s: i for i, s in enumerate(skills)}
m = (1 << n) - 1
d = {0: []}
for i, p in enumerate(people):
    mask = 0
    for s in p:
        mask |= 1 << idx[s]
    for old, team in list(d.items()):
        new = old | mask
        if new not in d or len(d[new]) > len(team) + 1:
            d[new] = team + [i]
print(d[m])
```

## 27 - Invert Binary Tree

```python
import ast
a = ast.literal_eval(input())
def inv(i):
    if i >= len(a) or a[i] is None:
        return
    l, r = 2 * i + 1, 2 * i + 2
    if l < len(a) and r < len(a):
        a[l], a[r] = a[r], a[l]
    inv(l)
    inv(r)
inv(0)
print(a)
```

## 35 - Trapping Rain Water

```python
a = list(map(int, input().strip("[]").split(",")))
l, r = 0, len(a) - 1
lm = rm = ans = 0
while l < r:
    if a[l] < a[r]:
        lm = max(lm, a[l])
        ans += lm - a[l]
        l += 1
    else:
        rm = max(rm, a[r])
        ans += rm - a[r]
        r -= 1
print(ans)
```

## 40 - Sudoku Validation

## 41 - Word Search in Grid

```python
import ast
g = ast.literal_eval(input())
words = ast.literal_eval(input())
m, n = len(g), len(g[0])
def find(w):
    def dfs(i, j, k):
        if k == len(w):
            return True
        if i < 0 or j < 0 or i >= m or j >= n or g[i][j] != w[k]:
            return False
        c = g[i][j]
        g[i][j] = "#"
        ok = dfs(i + 1, j, k + 1) or dfs(i - 1, j, k + 1) or dfs(i, j + 1, k + 1) or dfs(i, j - 1, k + 1)
        g[i][j] = c
        return ok
    return any(dfs(i, j, 0) for i in range(m) for j in range(n))
print([w for w in words if find(w)])
```

## 43 - Euler Path

```python
import ast
edges = ast.literal_eval(input())
g = {}
deg = {}
for a, b in edges:
    g.setdefault(a, []).append(b)
    g.setdefault(b, [])
    deg[a] = deg.get(a, 0) + 1
    deg[b] = deg.get(b, 0) - 1
start = next((x for x in deg if deg[x] == 1), edges[0][0] if edges else None)
st, path = [start], []
while st:
    v = st[-1]
    if g.get(v):
        st.append(g[v].pop())
    else:
        path.append(st.pop())
path.reverse()
print(path if len(path) == len(edges) + 1 else [])
```

## 49 - Cumulative Average Closure

```python
def avg():
    s = n = 0
    def add(x):
        nonlocal s, n
        s += x
        n += 1
        return s / n
    return add
f = avg()
while True:
    try:
        x = input()
        if not x:
            break
        print(f(float(x)))
    except EOFError:
        break
```

## 55 - Custom Iterator

```python
class Count:
    def __init__(self, n):
        self.n = n
        self.i = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.i > self.n:
            raise StopIteration
        x = self.i
        self.i += 1
        return x
for x in Count(int(input())):
    print(x)
```

## 62 - Rat in a Maze

```python
import ast
a = ast.literal_eval(input())
n = len(a)
r = []
def dfs(i, j, s):
    if i == n - 1 and j == n - 1:
        r.append(s)
        return
    a[i][j] = 0
    for di, dj, c in [(1, 0, "D"), (0, -1, "L"), (0, 1, "R"), (-1, 0, "U")]:
        x, y = i + di, j + dj
        if 0 <= x < n and 0 <= y < n and a[x][y] == 1:
            dfs(x, y, s + c)
    a[i][j] = 1
if n and a[0][0] and a[-1][-1]:
    dfs(0, 0, "")
print(sorted(r))
```

## 65 - Movie Ticket Class
